### Reverse string

```python
name = "Sanvi"
reverse = name[::-1]
print(reverse)
```

```python
city = "Mumbai"
reverse = city[::-1]
print(reverse)
```

### Palindrome 

```python
s = "Sneha"
if s == s[::-1]:
    print("palindrome")
else :
    print("Not palindrome")
```

```python
s = "madam"
if s == s[::-1]:
    print("palindrome")
else :
    print("Not palindrome")
```
